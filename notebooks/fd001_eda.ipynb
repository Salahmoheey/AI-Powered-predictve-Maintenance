# HexaMind – FD001 Exploratory Data Analysis

Quick EDA for raw NASA C-MAPSS FD001 training data.

```
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd()
if not (ROOT / "data" / "raw").exists():
    ROOT = ROOT.parent
DATA = ROOT / "data" / "raw"
columns = ["unit_id", "cycle", "setting_1", "setting_2", "setting_3"] + [f"sensor_{i}" for i in range(1, 22)]
train = pd.read_csv(DATA / "train_FD001.txt", sep=r"\s+", header=None, names=columns)
train.head()
```

```
print("Shape:", train.shape)
print("Engines:", train.unit_id.nunique())
print("Missing values:", int(train.isna().sum().sum()))
train.groupby("unit_id").cycle.max().describe()
```

```
sensor_columns = [c for c in train.columns if c.startswith("sensor_")]
train[sensor_columns].std().sort_values().plot(kind="bar", figsize=(12, 4), title="Sensor standard deviations")
plt.tight_layout()
plt.show()
```

```
train.groupby("unit_id").size().plot(kind="hist", bins=20, title="Cycles per engine")
plt.xlabel("Number of cycles")
plt.tight_layout()
plt.show()
```
